```python
# Method Defination
def finduniqueBST(n):
    n1, n2, sum_Val = 0, 0, 0
    # Base case condtion
    if n == 0 or n == 1:
        return 1
    # Catal Number series
    for i in range(1, n+1):
        n1 = finduniqueBST(i-1)
        n2 = finduniqueBST(n-i)
        sum_Val += n1*n2 
    return sum_Val
    
#Driver code

n = 10 # n indicates the number of nodes in the BST(unique)

result = finduniqueBST(n)
print("Number of unique possible BST:", result)
```
