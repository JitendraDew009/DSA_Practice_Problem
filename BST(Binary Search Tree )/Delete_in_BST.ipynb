```python
# Time Complexity : O(logn)
# class of the node in a BST
class Node:
    def __init__(self, data):
        self.left = None
        self.right = None
        self.data = data

# Insertion in BST
def insertBST(root, key):
    if root is None:
        return Node(key)
    else:
        if root.data == key:
            return root
        elif root.data < key:
            root.right = insertBST(root.right, key)
        else:
            root.left = insertBST(root.left, key)
    return root

# minimum value in BST
def minValBST(root):
    current = root
    # keep going left until you reach the last node
    while current.left is not None:
        current = current.left
    
    return current.data

# In-order traversal in BST
def inOrder(root):
    if root:
        inOrder(root.left)
        print(root.data, end=" ")
        inOrder(root.right)

# deletiton of node in BST
def deleteNode(root, key):
    # base case condition
    if root is None:
        return root
    # searching of the node to be deleted
    if key < root.data:
        root.left = deleteNode(root.left, key)

    elif key > root.data:
        root.right = deleteNode(root.right, key)
    # case 1 and 2
    else:
        if root.left is None:
            temp = root.right
            root = None
            return temp
        elif root.right is None:
            temp = root.left
            root = None
            return temp
        # case 3
        # inorder successor in the right subtree
        temp = minValBST(root.right)
        print()
        print("Min Val", temp)
        root.data = temp

        # Delete the inorder successor
        root.right = deleteNode(root.right, temp)
    return root



# Driver code
root = Node(100)
root = insertBST(root, 80)
root = insertBST(root, 110)
root = insertBST(root, 50)
root = insertBST(root, 90)

print("Inorder Traversal:", end=" ")
inOrder(root)
print("\n")
print("Minimum value in BST:", minValBST(root))

# no child # O(1)
deleteNode(root, 90)
print()
print("After Deletio of **90**, Inorder Traversal of the modified tree")
inOrder(root)

# Single child O(h)
deleteNode(root, 50)
print()
print("After Deletio of **50**, Inorder Traversal of the modified tree")
inOrder(root)

# two child O(h)
deleteNode(root, 100)
print()
print("After Deletio of **100**, Inorder Traversal of the modified tree")
inOrder(root)

```
