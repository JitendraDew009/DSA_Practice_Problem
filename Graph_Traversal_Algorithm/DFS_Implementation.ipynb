```python
# Time Complexity: O(V+E)

# Function defination
def depthFirstTraversal(visited, graph, node):
    if node not in visited:
        print(node, end=' ')
        visited.add(node)

        for adjacentNodes in graph[node]:
            if adjacentNodes not in visited:
                # recursion -> Store data structure to store
                # the function calls
                depthFirstTraversal(visited, graph, adjacentNodes)



# Driver Code
# Storing the data of the graph using the adjacency list
visited = set()
## Driected Graph
graph = {
    '5' : ['3', '7'],
    '3' : ['2', '4'],
    '7' : ['8'],
    '2' : [],
    '4' : ['8'],
    '8' : []
}
print("Depth First Traversal:")
depthFirstTraversal(visited, graph, '5')
```
